```python
from dotenv import load_dotenv
import os
from google import genai
from google.genai import types

load_dotenv(override=True)
gemini_api_key = os.getenv("GEMINI_API_KEY")
gemini_default_model = os.getenv("GEMINI_DEFAULT_MODEL", "").strip()

if not gemini_api_key:
    raise RuntimeError("GEMINI_API_KEY가 .env에 설정되어 있지 않습니다.")
if not gemini_default_model:
    raise RuntimeError("GEMINI_DEFAULT_MODEL이 .env에 설정되어 있지 않습니다.")

gemini_client = genai.Client(api_key=gemini_api_key)

def get_weather(city: str) -> dict:
    """도시의 현재 날씨를 반환한다. 실제 서비스에서는 날씨 API를 호출한다."""
    sample = {'서울': {'temperature': 21, 'condition': '맑음'}}
    return sample.get(city, {'error': '지원하지 않는 도시입니다.'})
chat = gemini_client.chats.create(
    model=gemini_default_model,
    config=types.GenerateContentConfig(tools=[get_weather]), #함수를 자동으로 분석해 등록하는 사용자 함수 도구
)
response = chat.send_message('서울 날씨를 알려줘.')
print(response.text)
```

```python
import yt_dlp
from openai import OpenAI
from pydub import AudioSegment

client = OpenAI()

def download_audio(youtube_url, output_path="temp_audio.%(ext)s"):
    """유튜브에서 오디오만 다운로드"""
    ydl_opts = {
        'format': 'bestaudio/best',
        'outtmpl': output_path,
        'postprocessors': [{
            'key': 'FFmpegExtractAudio',
            'preferredcodec': 'mp3',
            'preferredquality': '192',
        }],
    }
    with yt_dlp.YoutubeDL(ydl_opts) as ydl:
        ydl.download([youtube_url])
        info = ydl.extract_info(youtube_url, download=False)
        return ydl.prepare_filename(info).rsplit('.', 1)[0] + '.mp3'

def audio_to_text(audio_path):
    """Whisper API로 음성 → 텍스트"""
    with open(audio_path, "rb") as audio_file:
        transcript = client.audio.transcriptions.create(
            model="gpt-4o-mini-transcribe", #gpt-4o-transcribe
            file=audio_file,
            language="ko"  # 한국어 우선
        )
    return transcript.text
```

```python
from openai import OpenAI
client = OpenAI()
def text_extraction(text, prompt) :
    response = client.responses.create(
        model=default_model,
        input=[{'role': 'system' , 'content': prompt},
                {'role': 'user' , 'content': text}]
    )
    return response.output_text
```

```python
summary = text_extraction(full_text , summary_prompt)
print(summary)
key_points = text_extraction(full_text , key_points_prompt)
print(key_points)
action_items = text_extraction(full_text , action_items_prompt)
print(action_items)
sentiment = text_extraction(full_text , sentiment_prompt)
print(sentiment)
```
